```python
from TTL import *

N_start = 2
N_max = 30


for N in np.arange(N_start, N_max):
    MPO = Fourier_MPO(N, force=True)
    pre = [np.ndarray.flatten(M) for M in MPO]
    data = []
    for p in pre:
        data += list(p)
    Χ = [1] + [M.shape[-1] for M in MPO]
    print("Fourier_" + str(N))
    #Train_Shape(MPO, max=True)
    np.savez(r"Fourier_saved\Fourier_" + str(N), data, Χ, N)


for N in np.arange(N_start, N_max):
    MPO = Fourier_MPO(N, inv=True, force=True)
    pre = [np.ndarray.flatten(M) for M in MPO]
    data = []
    for p in pre:
        data += list(p)
    Χ = [1] + [M.shape[-1] for M in MPO]
    print("invFourier_" + str(N))
    #Train_Shape(MPO, max=True)
    np.savez(r"Fourier_saved\invFourier_" + str(N), data, Χ, N)
```
